# Gradio interface for Climate Risk and Economic Loss Prediction System

This system uses climate risk indicators to estimate economic losses caused by natural disasters. Enter the required values and click Submit to obtain the predicted economic loss.

```python
import gradio as gr

def predict_loss(
    cri_rank,
    cri_score,
    fatalities_per_100k_rank,
    fatalities_per_100k_total,
    fatalities_rank,
    fatalities_total,
    losses_per_gdp_rank,
    losses_per_gdp_total,
    losses_usdm_ppp_rank
):

    prediction = classifier.predict([[
        cri_rank,
        cri_score,
        fatalities_per_100k_rank,
        fatalities_per_100k_total,
        fatalities_rank,
        fatalities_total,
        losses_per_gdp_rank,
        losses_per_gdp_total,
        losses_usdm_ppp_rank
    ]])

    return prediction[0]

demo = gr.Interface(
    fn=predict_loss,

    inputs=[
        gr.Number(label="CRI Rank"),
        gr.Number(label="CRI Score"),
        gr.Number(label="Fatalities Per 100K Rank"),
        gr.Number(label="Fatalities Per 100K Total"),
        gr.Number(label="Fatalities Rank"),
        gr.Number(label="Fatalities Total"),
        gr.Number(label="Losses Per GDP Rank"),
        gr.Number(label="Losses Per GDP Total"),
        gr.Number(label="Losses USD PPP Rank")
    ],

    outputs=gr.Number(label="Predicted Economic Loss"),

    title="Climate Risk and Economic Loss Prediction",
    description="Enter climate risk indicators to predict economic losses."
)

demo.launch()
```
